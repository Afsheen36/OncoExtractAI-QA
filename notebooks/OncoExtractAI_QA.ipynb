```python
def extract_histology(text):
    text_lower = str(text).lower()

    if "adenocarcinoma" in text_lower:
        return "adenocarcinoma"
    elif "squamous cell carcinoma" in text_lower:
        return "squamous cell carcinoma"
    elif "non-small cell" in text_lower:
        return "non-small cell lung cancer"
    else:
        return "not found"

sample_df["histology_extracted"] = sample_df["text"].apply(extract_histology)

sample_df[["patient_filename", "histology_extracted"]].head()
```

```python
def extract_tumor_size(text):
    pattern = r'(\d+(\.\d+)?)\s*cm'
    match = re.search(pattern, str(text), re.IGNORECASE)

    if match:
        return match.group(0)
    else:
        return "not found"

sample_df["tumor_size_extracted"] = sample_df["text"].apply(extract_tumor_size)

sample_df[["patient_filename", "histology_extracted", "tumor_size_extracted"]].head()
```

```python
def extract_pT(text):
    pattern = r'\bpT[0-4][a-d]?\b'
    match = re.search(pattern, str(text), re.IGNORECASE)

    if match:
        return match.group(0)
    else:
        return "not found"

sample_df["pT_extracted"] = sample_df["text"].apply(extract_pT)

sample_df[["patient_filename", "histology_extracted", "tumor_size_extracted", "pT_extracted"]].head()
```

```python
def extract_pN(text):
    pattern = r'\bpN[0-3xX][a-d]?\b'
    match = re.search(pattern, str(text), re.IGNORECASE)

    if match:
        return match.group(0)
    else:
        return "not found"

sample_df["pN_extracted"] = sample_df["text"].apply(extract_pN)

sample_df[["patient_filename", "histology_extracted", "tumor_size_extracted", "pT_extracted", "pN_extracted"]].head()
```

```python
def split_sentences(text):
    sentences = re.split(r'(?<=[.!?])\s+', str(text))
    return sentences

sentences = split_sentences(sample_df.iloc[0]["text"])

sentences[:5]
```

```python
def find_evidence_sentence(text, value):
    if value == "not found":
        return "no evidence found"

    sentences = split_sentences(text)

    for sentence in sentences:
        if value.lower() in sentence.lower():
            return sentence

    return "no evidence found"

sample_df["histology_evidence"] = sample_df.apply(
    lambda row: find_evidence_sentence(row["text"], row["histology_extracted"]),
    axis=1
)

sample_df[["histology_extracted", "histology_evidence"]].head()
```

```python
sample_df["tumor_size_evidence"] = sample_df.apply(
    lambda row: find_evidence_sentence(row["text"], row["tumor_size_extracted"]),
    axis=1
)

sample_df[["tumor_size_extracted", "tumor_size_evidence"]].head()
```

```python
sample_df["pT_evidence"] = sample_df.apply(
    lambda row: find_evidence_sentence(row["text"], row["pT_extracted"]),
    axis=1
)

sample_df[["pT_extracted", "pT_evidence"]].head()
```

```python
sample_df["pN_evidence"] = sample_df.apply(
    lambda row: find_evidence_sentence(row["text"], row["pN_extracted"]),
    axis=1
)

sample_df[["pN_extracted", "pN_evidence"]].head()
```

```python
def qa_status(value, evidence):
    if value == "not found":
        return "manual review needed"
    elif evidence == "no evidence found":
        return "manual review needed"
    else:
        return "supported"

sample_df["histology_status"] = sample_df.apply(
    lambda row: qa_status(row["histology_extracted"], row["histology_evidence"]),
    axis=1
)

sample_df["tumor_size_status"] = sample_df.apply(
    lambda row: qa_status(row["tumor_size_extracted"], row["tumor_size_evidence"]),
    axis=1
)

sample_df["pT_status"] = sample_df.apply(
    lambda row: qa_status(row["pT_extracted"], row["pT_evidence"]),
    axis=1
)

sample_df["pN_status"] = sample_df.apply(
    lambda row: qa_status(row["pN_extracted"], row["pN_evidence"]),
    axis=1
)

sample_df[["histology_status", "tumor_size_status", "pT_status", "pN_status"]].head()
```
