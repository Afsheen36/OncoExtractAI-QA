```python
def extract_histology(text):
    text_lower = str(text).lower()

    if "adenocarcinoma" in text_lower:
        return "adenocarcinoma"
    elif "squamous cell carcinoma" in text_lower:
        return "squamous cell carcinoma"
    elif "non-small cell" in text_lower:
        return "non-small cell lung cancer"
    else:
        return "not found"

sample_df["histology_extracted"] = sample_df["text"].apply(extract_histology)

sample_df[["patient_filename", "histology_extracted"]].head()
```

```python
def extract_tumor_size(text):
    pattern = r'(\d+(\.\d+)?)\s*cm'
    match = re.search(pattern, str(text), re.IGNORECASE)

    if match:
        return match.group(0)
    else:
        return "not found"

sample_df["tumor_size_extracted"] = sample_df["text"].apply(extract_tumor_size)

sample_df[["patient_filename", "histology_extracted", "tumor_size_extracted"]].head()
```

```python
def extract_pT(text):
    pattern = r'\bpT[0-4][a-d]?\b'
    match = re.search(pattern, str(text), re.IGNORECASE)

    if match:
        return match.group(0)
    else:
        return "not found"

sample_df["pT_extracted"] = sample_df["text"].apply(extract_pT)

sample_df[["patient_filename", "histology_extracted", "tumor_size_extracted", "pT_extracted"]].head()
```

```python
def extract_pN(text):
    pattern = r'\bpN[0-3xX][a-d]?\b'
    match = re.search(pattern, str(text), re.IGNORECASE)

    if match:
        return match.group(0)
    else:
        return "not found"

sample_df["pN_extracted"] = sample_df["text"].apply(extract_pN)

sample_df[["patient_filename", "histology_extracted", "tumor_size_extracted", "pT_extracted", "pN_extracted"]].head()
```
